EX1.a)Simulaţi în Python experimentul de mai sus.

```python
import random


def simulation ():
    bila_rosie = False

    zar = random.randint(1,6)

    r,a,n=3,4,2
    if zar in {2,3,5}:
        n+=1
    elif zar==6:
        r+=1
    else:
        a+=1

    total_bile=r+a+n

    if random.random() < (r/total_bile):
        bila_rosie = True
    return bila_rosie

if simulation()==True:
    print ("bila rosie")
else:
    print("orice alta bila")
```

EX1.b)Folosind modelarea făcută, estimaţi probabilitatea de a obţine o bilă roşie.

```python
numar_simulari = 1000000
bile_rosii=0

for _ in range (numar_simulari):
    if simulation()==True:
        bile_rosii+=1

probabilitate_estimata = bile_rosii/numar_simulari 
print (probabilitate_estimata)
```
