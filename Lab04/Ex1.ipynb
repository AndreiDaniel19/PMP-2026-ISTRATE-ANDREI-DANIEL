# Exercitiu laborator 4 - Retele Markov

Variabile discrete binare: A1, A2, A3, A4, A5 in {0, 1}.
Muchii: (A1, A2), (A1, A3), (A2, A4), (A2, A5), (A3, A4), (A4, A5).
Potential pe clica maximala C: phi_C(a_C) = exp(beta * sum_{i,j in C, i < j} a_i * a_j).

## a) Graful retelei Markov si determinarea clicilor

Graful retelei Markov

Clicile modelului:
- Clici de marime 1 (noduri): {A1}, {A2}, {A3}, {A4}, {A5}
- Clici de marime 2 (muchii): {A1, A2}, {A1, A3}, {A2, A4}, {A2, A5}, {A3, A4}, {A4, A5}
- Clici de marime 3 (triunghi): {A2, A4, A5}

Clicile maximale ale retelei sunt:
1. C1 = {A2, A4, A5}
2. C2 = {A1, A2}
3. C3 = {A1, A3}
4. C4 = {A3, A4}

b)

```python
import itertools
import numpy as np


def energy(a1, a2, a3, a4, a5):
    return a1*a2 + a1*a3 + a2*a4 + a2*a5 + a3*a4 + a4*a5

beta = 1.0
states = list(itertools.product([0, 1], repeat=5))

weights = [np.exp(beta * energy(*s)) for s in states]
Z = sum(weights)
probs = [w / Z for w in weights]

max_p = max(probs)
max_states = [states[i] for i in range(len(states)) if np.isclose(probs[i], max_p)]

print("Z =", Z)
print("Probabilitate maxima =", max_p)
print("Stari de probabilitate maxima:", max_states)
print("\nPrimele 5 stari si probabilitatile lor:")
for i in range(5):
    print(f"P{states[i]} = {probs[i]:.6f}")
```
